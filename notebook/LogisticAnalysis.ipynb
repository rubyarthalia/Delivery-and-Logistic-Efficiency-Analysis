```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df_customer = pd.read_csv('olist_customers_dataset.csv');
df_customer.head()
```

```python
df_geolocation = pd.read_csv('olist_geolocation_dataset.csv');
df_geolocation.head()
```

```python
df_order_items = pd.read_csv('olist_order_items_dataset.csv');
df_order_items.head()
```

```python
df_order_payments = pd.read_csv('olist_order_payments_dataset.csv');
df_order_payments.head()
```

```python
df_order_reviews = pd.read_csv('olist_order_reviews_dataset.csv');
df_order_reviews.head()
```

```python
df_orders = pd.read_csv('olist_orders_dataset.csv');
df_orders.head()
```

```python
df_products = pd.read_csv('olist_products_dataset.csv');
df_products.head()
```

```python
df_sellers = pd.read_csv('olist_sellers_dataset.csv');
df_sellers.head()
```

```python
df_product_category_name_translation = pd.read_csv('product_category_name_translation.csv');
df_product_category_name_translation.head()
```

```python
print('Missing values in customer :' + str(df_customer.isnull().sum()))
print('\nMissing values in geolocation      :' + str(df_geolocation.isnull().sum()))
print('\nMissing values in order items     :' + str(df_order_items.isnull().sum()))
print('\nMissing values in order payments   :' + str(df_order_payments.isnull().sum()))
print('\nMissing values in order reviews    :' + str(df_order_reviews.isnull().sum()))
print('\nMissing values in orders           :' + str(df_orders.isnull().sum()))
print('\nMissing values in products         :' + str(df_products.isnull().sum()))
print('\nMissing values in sellers          :' + str(df_sellers.isnull().sum()))
print('\nMissing values in category translations :' + str(df_product_category_name_translation.isnull().sum()))
```

```python
# There's a lot of missing values in the review comment title and message, so we can drop these columns. 
# And since we are analyzing the logistics of the orders, we can also drop the review score, as it is not relevant to our analysis.
df_order_reviews.drop(columns=['review_comment_title', 'review_comment_message'], inplace=True)
```

```python
# We need to drop the rows with missing values in the order_delivered_customer_date and order_delivered_carrier_date columns
# as these are the columns we will be analyzing.
df_orders.dropna(subset=['order_delivered_customer_date', 'order_delivered_carrier_date'], inplace=True)
```

```python
# We fill the missing values in the product category name with 'unknown' as it is a categorical variable 
# and we don't want to drop the rows with missing values as it would reduce our dataset significantly.
df_products['product_category_name'].fillna('unknown', inplace=True)
```

```python
# We fill the missing values in the product weight, length, height and width with the median value of each column as they are numerical variables 
# and we don't want to drop the rows with missing values as it would reduce our dataset significantly.
for col in ['product_weight_g', 'product_length_cm', 'product_height_cm', 'product_width_cm']:
    df_products[col].fillna(df_products[col].median(), inplace=True)
```

```python
# We can drop the product name length, description length and photos quantity columns as they are not relevant to our analysis.
df_products.drop(columns=['product_name_lenght', 'product_description_lenght', 'product_photos_qty'], inplace=True)
```

```python
print('Duplicates in customer :' + str(df_customer.duplicated().sum()))
print('Duplicates in geolocation :' + str(df_geolocation.duplicated().sum()))
print('Duplicates in order items :' + str(df_order_items.duplicated().sum()))   
print('Duplicates in order payments :' + str(df_order_payments.duplicated().sum()))
print('Duplicates in order reviews :' + str(df_order_reviews.duplicated().sum()))   
print('Duplicates in orders :' + str(df_orders.duplicated().sum()))
print('Duplicates in products :' + str(df_products.duplicated().sum())) 
print('Duplicates in sellers :' + str(df_sellers.duplicated().sum()))
print('Duplicates in category translations :' + str(df_product_category_name_translation.duplicated().sum()))   
```

```python
# There are a lot of duplicates in the geolocation dataset, but we can drop them and only keep the first occurrence of each geolocation zip code prefix 
# as it is not relevant to our analysis to have multiple rows with the same geolocation zip code prefix.
df_geolocation.drop_duplicates(subset=['geolocation_zip_code_prefix'], keep='first', inplace=True)
```

```python
df_order_items['shipping_limit_date'] = pd.to_datetime(df_order_items['shipping_limit_date'])
df_orders['order_purchase_timestamp'] = pd.to_datetime(df_orders['order_purchase_timestamp'])
df_orders['order_approved_at'] = pd.to_datetime(df_orders['order_approved_at'])
df_orders['order_delivered_carrier_date'] = pd.to_datetime(df_orders['order_delivered_carrier_date'])
df_orders['order_delivered_customer_date'] = pd.to_datetime(df_orders['order_delivered_customer_date'])
df_orders['order_estimated_delivery_date'] = pd.to_datetime(df_orders['order_estimated_delivery_date'])
```

```python
# orders + order_items
df_combined = df_orders.merge(df_order_items, on='order_id', how='left')

# + products
df_combined = df_combined.merge(df_products, on='product_id', how='left')

# + category translation (get English names)
df_combined = df_combined.merge(df_product_category_name_translation, on='product_category_name', how='left')

# + sellers
df_combined = df_combined.merge(df_sellers, on='seller_id', how='left')

# + customers
df_combined = df_combined.merge(df_customer, on='customer_id', how='left')

# + order_reviews (only review_score)
df_combined = df_combined.merge(
    df_order_reviews[['order_id', 'review_score']], 
    on='order_id', 
    how='left'
)

# + order_payments (aggregate first to avoid row explosion)
payments_agg = df_order_payments.groupby('order_id').agg(
    total_payment=('payment_value', 'sum'),
    payment_type=('payment_type', 'first'),
    payment_installments=('payment_installments', 'max')
).reset_index()
df_combined = df_combined.merge(payments_agg, on='order_id', how='left')

# Check result
print(df_combined.shape)
print(df_combined.columns.tolist())
```

```python
# Check remaining nulls in key columns
key_cols = [
    'order_delivered_customer_date',
    'order_estimated_delivery_date',
    'review_score',
    'price',
    'freight_value',
    'seller_state',
    'customer_state',
    'product_category_name_english'
]
print(df_combined[key_cols].isnull().sum())

# Check order_status distribution
print(df_combined['order_status'].value_counts())

# Drop columns that won't use
df_combined.drop(columns=[
    'order_approved_at',
    'customer_zip_code_prefix',
    'seller_zip_code_prefix',
    'product_category_name',  # keeping only english version
    'customer_id',            # keeping customer_unique_id instead
], inplace=True)
print("Shape after dropping columns:", df_combined.shape)
```

```python
df_combined['review_score'].fillna(df_combined['review_score'].median(), inplace=True)
```

```python
df_combined['product_category_name_english'].fillna('unknown', inplace=True)
```

```python
df_combined = df_combined[df_combined['order_status'] == 'delivered'].copy()
df_combined.drop(columns=['order_status'], inplace=True)
```

```python
df_combined.to_csv('combined_data.csv', index=False)
print("Final shape:", df_combined.shape)
```

```python
# Delivery feature engineering

# Delivery delay (positive = late, negative = early)
df_combined['delivery_delay_days'] = (
    df_combined['order_delivered_customer_date'] - 
    df_combined['order_estimated_delivery_date']
).dt.days

# Actual delivery time (purchase → delivered to customer)
df_combined['actual_delivery_days'] = (
    df_combined['order_delivered_customer_date'] - 
    df_combined['order_purchase_timestamp']
).dt.days

# Processing time (purchase → shipped to carrier)
df_combined['processing_time_days'] = (
    df_combined['order_delivered_carrier_date'] - 
    df_combined['order_purchase_timestamp']
).dt.days

# On-time flag
df_combined['is_late'] = (df_combined['delivery_delay_days'] > 0).astype(int)

# Quick  check
print(df_combined[['delivery_delay_days', 'actual_delivery_days', 
                   'processing_time_days', 'is_late']].describe())

print(f"\nLate orders : {df_combined['is_late'].sum():,}")
print(f"On-time orders: {(df_combined['is_late'] == 0).sum():,}")
print(f"On-time rate  : {(df_combined['is_late'] == 0).mean()*100:.1f}%")
```

```python
# Check negative processing times - these are data quality issues
print("Negative processing times:", (df_combined['processing_time_days'] < 0).sum())

# Filter them out - physically impossible
df_combined = df_combined[df_combined['processing_time_days'] >= 0].copy()
print("Shape after filter:", df_combined.shape)
```

```python
# Delivery delay distribution
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram of delay days
axes[0].axvline(0, color='red', linestyle='--', linewidth=1.5, label='On-time threshold')
sns.histplot(df_combined['delivery_delay_days'], bins=60, ax=axes[0], color='steelblue')
axes[0].set_title('Delivery Delay Distribution')
axes[0].set_xlabel('Days (negative = early, positive = late)')
axes[0].legend()

# On-time vs late pie
labels = ['On-time / Early (93.4%)', 'Late (6.6%)']
sizes  = [103539, 7292]
colors = ['#2196F3', '#EF5350']
axes[1].pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)
axes[1].set_title('On-time Delivery Rate')

plt.tight_layout()
plt.savefig('delivery_overview.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Late orders by seller state
late_by_state = df_combined[df_combined['is_late'] == 1].groupby('seller_state').size()
total_by_state = df_combined.groupby('seller_state').size()
late_rate_by_state = (late_by_state / total_by_state * 100).dropna().sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 5))
sns.barplot(x=late_rate_by_state.values, y=late_rate_by_state.index, palette='Reds_r')
plt.title('Top 10 Seller States by Late Delivery Rate (%)')
plt.xlabel('Late Delivery Rate (%)')
plt.ylabel('Seller State')
plt.tight_layout()
plt.savefig('late_by_state.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Delay vs review score 
delay_vs_review = df_combined.groupby('review_score')['delivery_delay_days'].mean().reset_index()

plt.figure(figsize=(8, 5))
sns.barplot(data=delay_vs_review, x='review_score', y='delivery_delay_days', palette='coolwarm')
plt.axhline(0, color='black', linestyle='--', linewidth=1)
plt.title('Average Delivery Delay by Review Score')
plt.xlabel('Review Score (1–5)')
plt.ylabel('Avg Delay Days (negative = early)')
plt.tight_layout()
plt.savefig('delay_vs_review.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Monthly revenue trend
df_combined['order_month'] = df_combined['order_purchase_timestamp'].dt.to_period('M')
monthly_revenue = df_combined.groupby('order_month')['price'].sum().reset_index()
monthly_revenue['order_month'] = monthly_revenue['order_month'].astype(str)

plt.figure(figsize=(14, 5))
plt.plot(monthly_revenue['order_month'], monthly_revenue['price'], 
         color='steelblue', linewidth=2, marker='o', markersize=3)
plt.xticks(rotation=45, ha='right', fontsize=8)
plt.title('Monthly Revenue Trend (2016–2018)')
plt.xlabel('Month')
plt.ylabel('Total Revenue (BRL)')
# Annotate the Nov 2017 Black Friday spike
plt.annotate('Black Friday\nNov 2017', 
             xy=('2017-11', monthly_revenue[monthly_revenue['order_month']=='2017-11']['price'].values[0]),
             xytext=('2017-08', monthly_revenue['price'].max() * 0.9),
             arrowprops=dict(arrowstyle='->', color='red'),
             fontsize=9, color='red')
plt.tight_layout()
plt.savefig('monthly_revenue.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Top 10 product categories by revenue
top_categories = (df_combined.groupby('product_category_name_english')['price']
                  .sum()
                  .sort_values(ascending=False)
                  .head(10))

plt.figure(figsize=(10, 6))
sns.barplot(x=top_categories.values, y=top_categories.index, palette='Blues_r')
plt.title('Top 10 Product Categories by Revenue (BRL)')
plt.xlabel('Total Revenue (BRL)')
plt.ylabel('')
plt.tight_layout()
plt.savefig('top_categories.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Freight cost vs product price 
df_combined['freight_ratio'] = df_combined['freight_value'] / df_combined['price']

plt.figure(figsize=(8, 5))
sns.scatterplot(data=df_combined.sample(5000, random_state=42), 
                x='price', y='freight_value', 
                alpha=0.3, color='steelblue', s=10)
plt.axline((0,0), slope=df_combined['freight_ratio'].median(), 
           color='red', linestyle='--', linewidth=1.5,
           label=f'Median freight ratio: {df_combined["freight_ratio"].median():.1%}')
plt.title('Freight Cost vs Product Price')
plt.xlabel('Product Price (BRL)')
plt.ylabel('Freight Value (BRL)')
plt.legend()
plt.xlim(0, 500)
plt.ylim(0, 100)
plt.tight_layout()
plt.savefig('freight_vs_price.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Payment type breakdown 
payment_counts = df_combined['payment_type'].value_counts()

plt.figure(figsize=(7, 5))
sns.barplot(x=payment_counts.values, y=payment_counts.index, palette='Blues_r')
plt.title('Orders by Payment Type')
plt.xlabel('Number of Orders')
plt.ylabel('')
plt.tight_layout()
plt.savefig('payment_types.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Top 10 customer states by order volume 
top_customer_states = (df_combined.groupby('customer_state')['order_id']
                       .nunique()
                       .sort_values(ascending=False)
                       .head(10))

plt.figure(figsize=(10, 5))
sns.barplot(x=top_customer_states.values, y=top_customer_states.index, palette='Blues_r')
plt.title('Top 10 Customer States by Order Volume')
plt.xlabel('Number of Orders')
plt.ylabel('State')
plt.tight_layout()
plt.savefig('customer_states.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Review score distribution
review_dist = df_combined['review_score'].value_counts().sort_index()

plt.figure(figsize=(8, 5))
sns.barplot(x=review_dist.index, y=review_dist.values, palette='RdYlGn')
plt.title('Review Score Distribution')
plt.xlabel('Review Score')
plt.ylabel('Number of Orders')
for i, v in enumerate(review_dist.values):
    plt.text(i, v + 200, f'{v/len(df_combined)*100:.1f}%', ha='center', fontsize=9)
plt.tight_layout()
plt.savefig('review_distribution.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
# Avg review score by product category (top 10)
avg_review_by_cat = (df_combined.groupby('product_category_name_english')['review_score']
                     .mean()
                     .sort_values(ascending=False)
                     .head(10))

plt.figure(figsize=(10, 6))
sns.barplot(x=avg_review_by_cat.values, y=avg_review_by_cat.index, palette='Greens_r')
plt.axvline(df_combined['review_score'].mean(), color='red', linestyle='--', 
            linewidth=1.5, label=f'Overall avg: {df_combined["review_score"].mean():.2f}')
plt.title('Top 10 Categories by Avg Review Score')
plt.xlabel('Average Review Score')
plt.ylabel('')
plt.xlim(3.5, 5)
plt.legend()
plt.tight_layout()
plt.savefig('review_by_category.png', dpi=150, bbox_inches='tight')
plt.show()
```
